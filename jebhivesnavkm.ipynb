```python
def func(n):
    a = [0, 1]
    for i in range(1, n):
        a.append(a[i]+a[i-1])

    return a
try:
    print(func(abs('-6.2')))
except TypeError:
    try:
        n = int(abs('-6.2'))
        print(func(n))
    except ValueError as v:
        print(v)
    
```
